```python
s = input("Enter a sentence: ")
words = s.split()
result = []

for word in words:
    rev = ""
    for ch in word:
        rev = ch + rev
    result.append(rev)

print(" ".join(result))
```

```python
s = input("Enter a string: ")
checked = ""

for ch in s:
    if ch not in checked:
        freq = 0
        for c in s:
            if c == ch:
                freq += 1

        print(ch, ":", freq)
        checked += ch
```

```python
s = input("Enter a string: ")
found = False

for ch in s:
    freq = 0

    for c in s:
        if c == ch:
            freq += 1

    if freq == 1:
        print("First non-repeating character:", ch)
        found = True
        break

if not found:
    print("No non-repeating character")
```

```python
s = input("Enter a string: ")
seen = ""

for ch in s:
    if ch in seen:
        print("First repeating character:", ch)
        break
    seen += ch
else:
    print("No repeating character")
```

```python
s1 = input("Enter first string: ").lower()
s2 = input("Enter second string: ").lower()

if len(s1) != len(s2):
    print("Not anagrams")
else:
    for ch in s1:
        if s1.count(ch) != s2.count(ch):
            print("Not anagrams")
            break
    else:
        print("Anagrams")
```

```python
s = input("Enter a string: ")
letters = ""
digits = ""

for ch in s:
    if ch.isdigit():
        digits += ch
    else:
        letters += ch

print("Result:", letters + digits)
```

```python
s = input("Enter an expression: ")
stack = []
pairs = {')': '(', ']': '[', '}': '{'}
balanced = True

for ch in s:
    if ch in "([{":
        stack.append(ch)

    elif ch in ")]}":
        if not stack or stack[-1] != pairs[ch]:
            balanced = False
            break
        stack.pop()

if balanced and not stack:
    print("Balanced parentheses")
else:
    print("Not balanced")
```
