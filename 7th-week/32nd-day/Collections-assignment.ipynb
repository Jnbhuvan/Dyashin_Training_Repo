```python
# 29. Find the longest consecutive increasing sequence in a tuple. Return the sequence, not just its length.
numbers = (1, 2, 5, 6, 7, 3, 4, 8)

if not numbers:
    longest_seq = ()
else:
    current_seq = [numbers[0]]
    longest_seq = [numbers[0]]

    for i in range(1, len(numbers)):
        if numbers[i] == numbers[i - 1] + 1:
            current_seq.append(numbers[i])
        else:
            if len(current_seq) > len(longest_seq):
                longest_seq = current_seq
            current_seq = [numbers[i]]

    if len(current_seq) > len(longest_seq):
        longest_seq = current_seq

longest_seq = tuple(longest_seq)
print("Longest increasing sequence:", longest_seq)
print("Length:", len(longest_seq))
```

```python
# 90. Given employee records as tuples, identify duplicate employee IDs using a set and create a dictionary of valid records. Keep the first record for each ID.
employees = ((101, 'Asha'), (102, 'Rahul'), (101, 'Asha R'), (103, 'Meena'))
seen_ids = set()
duplicate_ids = set()
valid_records = {}
for emp_id, name in employees:
    if emp_id in seen_ids:
        duplicate_ids.add(emp_id)
    else:
        seen_ids.add(emp_id)
        valid_records[emp_id] = name
print("Duplicate IDs:", duplicate_ids)
print("Valid records:", valid_records)
```

```python
# 91. Given two dictionaries of marks, find students who appeared in both exams and calculate combined totals.
exam1 = {'Asha': 80, 'Rahul': 70, 'Meena': 90}
exam2 = {'Rahul': 75, 'Meena': 85, 'Kiran': 88}
common_students = {}
for student in exam1:
    if student in exam2:
        common_students[student] = exam1[student] + exam2[student]
print("Common students:", common_students)
```
