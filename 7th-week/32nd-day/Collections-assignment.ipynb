```python
# 12. Find the second-largest unique number in a tuple.
numbers = (12, 45, 32, 45, 28, 32, 40)

first = float('-inf')
second = float('-inf')

for num in set(numbers):
    if num > first:
        second = first
        first = num
    elif num > second:
        second = num

print("Second-largest unique number:", second)
```

```python
# 19. Sort a tuple of numbers in ascending order without using sorted().
numbers = (35, 12, 48, 7, 23)

num_list = list(numbers)
for i in range(len(num_list)):
    for j in range(len(num_list) - i - 1):
        if num_list[j] > num_list[j + 1]:
            num_list[j], num_list[j + 1] = num_list[j + 1], num_list[j]

sorted_tuple = tuple(num_list)
print(sorted_tuple)
```

```python
# 21. For student marks, calculate the average, highest mark, lowest mark, and number of students who passed (marks >= 35).
marks = (85, 32, 92, 67, 28, 75)

total = 0
highest = marks[0]
lowest = marks[0]
passed_count = 0

for m in marks:
    total += m
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
    if m >= 35:
        passed_count += 1

average = total / len(marks)

print(f"Average: {average:.2f}")
print("Highest:", highest)
print("Lowest:", lowest)
print("Passed:", passed_count)
```

```python
# 29. Find the longest consecutive increasing sequence in a tuple. Return the sequence, not just its length.
numbers = (1, 2, 5, 6, 7, 3, 4, 8)

if not numbers:
    longest_seq = ()
else:
    current_seq = [numbers[0]]
    longest_seq = [numbers[0]]

    for i in range(1, len(numbers)):
        if numbers[i] == numbers[i - 1] + 1:
            current_seq.append(numbers[i])
        else:
            if len(current_seq) > len(longest_seq):
                longest_seq = current_seq
            current_seq = [numbers[i]]

    if len(current_seq) > len(longest_seq):
        longest_seq = current_seq

longest_seq = tuple(longest_seq)
print("Longest increasing sequence:", longest_seq)
print("Length:", len(longest_seq))
```

```python
# 30. Given daily temperatures, identify the days when temperature was above the average.
temperatures = (30, 32, 29, 35, 31)

average = sum(temperatures) / len(temperatures)
print(f"Average: {average}")

above_avg_info = []
for i in range(len(temperatures)):
    if temperatures[i] > average:
        above_avg_info.append(f"Day {i + 1} ({temperatures[i]})")

print("Days above average:", ", ".join(above_avg_info))
```
