```python
# Apply a 10% discount to product prices greater than ₹1,000. Leave other prices unchanged. 
prices = [500, 1200, 2500, 900, 1500]

discounted_prices = [
    price * 0.9 if price > 1000 else price
    for price in prices
]

print(discounted_prices)
```

```python
 # Find the temperatures that are above the average temperature. 
temperatures = [28, 30, 32, 26, 34] 


average = sum(temperatures) / len(temperatures)

above_average = [temp for temp in temperatures if temp > average]

print(above_average)
```

```python
# . Find the salaries greater than the median salary. For an even number of salaries, use the average
# of the two middle values after sorting.
salaries = [30000, 45000, 25000, 60000, 40000]


sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 0:
    median = (sorted_salaries[n // 2 - 1] + sorted_salaries[n // 2]) / 2
else:
    median = sorted_salaries[n // 2]

above_median = [salary for salary in salaries if salary > median]

print("Median salary:", median)
print("Salaries greater than median:", above_median)
```

```python
quantities = [2, 3, 1]
prices = [100, 50, 250]

total_bill = sum(q * p for q, p in zip(quantities, prices))

print("Total bill:", total_bill)
```

```python
scores = [78, 95, 88, 95, 72, 91, 88, 84]

top_three = []

while len(top_three) < 3:
    highest = max(scores)

    if highest not in top_three:
        top_three.append(highest)

    scores = [s for s in scores if s != highest]

print("Top three scores:", top_three)
```

```python
numbers = [1, 2, 5, 3, 4, 6, 2, 8]

longest_run = []
current_run = [numbers[0]]

for i in range(1, len(numbers)):
    if numbers[i] > numbers[i - 1]:
        current_run.append(numbers[i])
    else:
        if len(current_run) > len(longest_run):
            longest_run = current_run
        current_run = [numbers[i]]

if len(current_run) > len(longest_run):
    longest_run = current_run

print("Longest increasing run:", longest_run)
```

```python
numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = numbers[0]
current_sum = numbers[0]

start = 0
end = 0
temp_start = 0

for i in range(1, len(numbers)):
    if current_sum < 0:
        current_sum = numbers[i]
        temp_start = i
    else:
        current_sum += numbers[i]

    if current_sum > max_sum:
        max_sum = current_sum
        start = temp_start
        end = i

sublist = numbers[start:end + 1]

print("Maximum sublist sum:", max_sum)
print("Sublist:", sublist)
```
