```python
# . Find the salaries greater than the median salary. For an even number of salaries, use the average
# of the two middle values after sorting.
salaries = [30000, 45000, 25000, 60000, 40000]


sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 0:
    median = (sorted_salaries[n // 2 - 1] + sorted_salaries[n // 2]) / 2
else:
    median = sorted_salaries[n // 2]

above_median = [salary for salary in salaries if salary > median]

print("Median salary:", median)
print("Salaries greater than median:", above_median)
```

```python
scores = [78, 95, 88, 95, 72, 91, 88, 84]

top_three = []

while len(top_three) < 3:
    highest = max(scores)

    if highest not in top_three:
        top_three.append(highest)

    scores = [s for s in scores if s != highest]

print("Top three scores:", top_three)
```

```python
numbers = [1, 2, 5, 3, 4, 6, 2, 8]

longest_run = []
current_run = [numbers[0]]

for i in range(1, len(numbers)):
    if numbers[i] > numbers[i - 1]:
        current_run.append(numbers[i])
    else:
        if len(current_run) > len(longest_run):
            longest_run = current_run
        current_run = [numbers[i]]

if len(current_run) > len(longest_run):
    longest_run = current_run

print("Longest increasing run:", longest_run)
```

```python
numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = numbers[0]
current_sum = numbers[0]

start = 0
end = 0
temp_start = 0

for i in range(1, len(numbers)):
    if current_sum < 0:
        current_sum = numbers[i]
        temp_start = i
    else:
        current_sum += numbers[i]

    if current_sum > max_sum:
        max_sum = current_sum
        start = temp_start
        end = i

sublist = numbers[start:end + 1]

print("Maximum sublist sum:", max_sum)
print("Sublist:", sublist)
```

```python
numbers = [5, 2, 9, 1, 5, 6]

# Implementing Bubble Sort
for i in range(len(numbers)):
    for j in range(len(numbers) - i - 1):
        if numbers[j] > numbers[j + 1]:
            # Swap
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Ascending:", numbers)
```

```python
numbers = [12, 4, 19, 7, 2]

# Implementing Bubble Sort for descending order
for i in range(len(numbers)):
    for j in range(len(numbers) - i - 1):
        if numbers[j] < numbers[j + 1]:
            # Swap
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Descending:", numbers)
```

```python
numbers = [10, 5, 8, 10, 7, 8]

first_largest = float("-inf")
second_largest = float("-inf")

for num in numbers:
    if num > first_largest:
        second_largest = first_largest
        first_largest = num
    elif num > second_largest and num != first_largest:
        second_largest = num

print("Second-largest distinct value:", second_largest)
```

```python
numbers = [12, 5, 8, 12, 9, 3]

first = float("-inf")
second = float("-inf")
third = float("-inf")

for num in numbers:
    if num > first:
        third = second
        second = first
        first = num
    elif num > second:
        third = second
        second = num
    elif num > third:
        third = num

print("Third-largest element:", third)
```

```python
numbers = [2, 4, 3, 5, 7, 8, 1]
target = 9

seen = set()
pairs = []
used_indices = set()

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if (
            numbers[i] + numbers[j] == target
            and i not in used_indices
            and j not in used_indices
        ):
            pairs.append((numbers[i], numbers[j]))
            used_indices.add(i)
            used_indices.add(j)
            break

print("Pairs:", pairs)
```

```python
numbers = [-1, 0, 1, 2, -1, -4]

numbers.sort()
triplets = set()

for i in range(len(numbers) - 2):
    left = i + 1
    right = len(numbers) - 1
    while left < right:
        current_sum = numbers[i] + numbers[left] + numbers[right]
        if current_sum == 0:
            triplets.add((numbers[i], numbers[left], numbers[right]))
            left += 1
            right -= 1
        elif current_sum < 0:
            left += 1
        else:
            right -= 1

print("Triplets:", list(triplets))
```

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

row_sums = []
for row in matrix:
    row_sums.append(sum(row))

column_sums = []
for col in range(len(matrix[0])):
    col_sum = 0
    for row in range(len(matrix)):
        col_sum += matrix[row][col]
    column_sums.append(col_sum)

print("Row sums:", row_sums)
print("Column sums:", column_sums)
```

```python
matrix = [[1, 2, 3], [10, 1, 1], [4, 5, 6]]

max_sum = float("-inf")
row_number = -1
largest_row = []

for idx, row in enumerate(matrix):
    current_sum = sum(row)
    if current_sum > max_sum:
        max_sum = current_sum
        row_number = idx + 1
        largest_row = row

print("Row number:", row_number)
print("Row:", largest_row)
print("Row sum:", max_sum)
```
