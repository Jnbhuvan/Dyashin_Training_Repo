```python
t = (1,2,3,4,5)
print(t[2])
print(t[-4])
print(t[1:3])
print(t[::-1])
print(t[::2])
print(t.count(3))
print(t.index(3))
for i in t:
    print(i)
    
```

```python
person = ("Jainag-Bhuvan", 22, "male")  # Packing
name,age,gender = person                # Unpacking

print(name)
print(age)
print(person)
```

```python
set = {1,2,3,4,5,6}
print(set.add(23))
print(set.update([12,34,56]))
print(set.remove(12))
# print(set.remove(30))
print(set.discard(34))
print(set.discard(40))
print(set)
```

```python
a = {1,2,3,4,5}
b = {2,3,4,5,6,7,8}
print(a.union(b))
print(a&b)
print(a-b)
print(a^b)
```

```python
std = {
    "name":"mary",
    "age":23,
    "gender":"female"
}

print(std)
print(std.keys())
print(std.values())
print(std.items())
print(std.update({"marks":78}))
print(std.pop("age"))
print(std.popitem())

print(std.copy())

for k,v in std.items():
    print(k,v)

print(std)
```
